```python
def process_node(node, graph):    
    dependencies = list(graph.predecessors(node))
    if not dependencies:
        # Node has no dependencies, process it
        print(f"Processing node {node}")
        # Add processing logic here
        return node
    else:
        # Node has dependencies, wait for them to complete
        for dependency in dependencies:
            process_node((dependency, graph))
```
