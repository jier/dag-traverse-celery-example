```python
import networkx as nx
from graphlib import TopologicalSorter as ts
from collections import deque, defaultdict
import random
import math
```

```python
make_dependency_0 = {
    1: [3],
    2: [4],
    3: [5],
    4: [5],
    5: [6, 7],
    6: [8],
    7: [8]
}
make_dependency_1 =dict([
    (1, [3]),
    (2, [3]),
    (3, [4, 5]),
    (4, [6]),
    (5, [6]),
    (6, [7])
])

make_dependency_2 = {
    1: [3],
    2: [3],
    3: [4],
    4: [5]
}

graph = {"D": {"B", "C"}, "C": {"A"}, "B": {"A"}}
graph_2 = {"1": {"3"}, "2": {"3"}, "3": {"4","5"}, "4":{}, "5":{"6"}, "6":{}}
# Create a directed graph for each adjacency list
G0 = nx.DiGraph()
G1 = nx.DiGraph()
G2 = nx.DiGraph()
G3 = nx.DiGraph()
G4  = nx.DiGraph()
# Add edges to the graphs
for node, successors in make_dependency_1.items():
    for successor in successors:
        G1.add_edge(node, successor)

for node, successors in make_dependency_2.items():
    for successor in successors:
        G2.add_edge(node, successor)
# for node, successors in graph.items():
#     for successor in successors:
#         G4.add_edge(node, successor)
d = make_dependency_0
for node in d.keys():
    # print('node:', node)
    nodes = d[node]
    # print('nodes:', nodes)
    if len(nodes) == 0:
        G0.add_node(int(node))
        continue
    G0.add_edges_from([(int(node), n) for n in nodes])
d = make_dependency_1
for node in d.keys():
    # print('node:', node)
    nodes = d[node]
    # print('nodes:', nodes)
    if len(nodes) == 0:
        G3.add_node(int(node))
        continue
    G3.add_edges_from([(int(node), n) for n in nodes])

for node in graph.keys():
    # print('node:', node)
    nodes = graph[node]
    # print('nodes:', nodes)
    if len(nodes) == 0:
        G4.add_node(node)
        continue
    G4.add_edges_from([(node, n) for n in nodes])
```

```python
if nx.is_directed_acyclic_graph(G3):
    print('G3 is a DAG')
else: 
    print('G3 is not a DAG')    
if nx.is_directed_acyclic_graph(G4):
    print('G4 is a DAG')
else: 
    print('G4 is not a DAG')  
print(list(nx.topological_sort(G0)))
print(list(nx.topological_sort(G4)))
print(list(nx.topological_sort(nx.line_graph(G4))))
print(list(nx.topological_sort(G3)))
print(list(nx.topological_sort(nx.line_graph(G3))))
print(list(ts(graph).static_order()))
print(list(ts(graph_2).static_order()))
```

```python
# Find nodes with no dependencies (i.e., nodes with in-degree 0)
# Find nodes with no predecessors
no_predecessors = [node for node in G1.nodes() if len(list(G1.predecessors(node))) == 0]

print("Nodes with no predecessors:{}".format(no_predecessors))

visited = set()
queue = deque(no_predecessors)
while queue:
    node = queue.popleft()
    print("Starting node: {}".format(node))
    if node not in visited:
        print(f"Processing node {node} in parallel...")
        # process_node.apply_async((node, G))
        # add visited node to the set
        visited.add(node)
        # Add nodes with no predecessors to the queue
        for neighbor in G1.neighbors(node):
            print('Look for neighbors in for-loop')
            if neighbor not in visited:
                if  not G1.in_degree(neighbor) == 0:
                    print('No predecessor in neighbour, append to process neighbour: {}'.format(neighbor))
                    queue.append(neighbor)
            
print('DONE?')
```

```python
def process_node(node, graph):    
    dependencies = list(graph.predecessors(node))
    if not dependencies:
        # Node has no dependencies, process it
        print(f"Processing node {node}")
        # Add processing logic here
        return node
    else:
        # Node has dependencies, wait for them to complete
        for dependency in dependencies:
            process_node((dependency, graph))
```
